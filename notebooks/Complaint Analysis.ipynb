```python
# prompt: list columns with more than 70 % nan

nan_counts = df.isnull().sum()
nan_percentage = nan_counts / len(df)
columns_to_drop = nan_percentage[nan_percentage > 0.1].index.tolist()
columns_to_drop
```

Consumer disputed filled null values  with "No"

droped Tags column since it is more than 70 % nan and it is not essenscial  

```python
# prompt: drop row with null "Issue"

df.dropna(subset=["Issue"], inplace=True)
```

```python
df = df.drop(columns=['Tags'])
df = df.drop(columns=['Sub-issue'])
df = df.drop(columns=['Sub-product'])
```

```python
df['Consumer disputed?'] = df['Consumer disputed?'].fillna('No')
df['Consumer consent provided?'] = df['Consumer consent provided?'].fillna('No consent provided')
df['Company public response'] = df['Company public response'].fillna('No public response')
df['State'] = df['State'].fillna('unknown')
df['ZIP code'] = df['ZIP code'].fillna('unknown')
df['Company response to consumer'] = df['Company response to consumer'].fillna('No response')

```

for null of Consumer complaint narrative Imputed with Issue + Sub-issue

```python
df['Consumer complaint narrative'] = df['Consumer complaint narrative'].fillna(
    df['Issue']
)
```
